```python
import pandas as pd
import os

# Create target directory for clean data if it doesn't exist
os.makedirs('../data/processed', exist_ok=True)
print("Environment ready.")
```

```python
print("Loading ridership data...")
df_ridership = pd.read_csv('../data/raw/cta_ridership.csv')

# 1. Clean the 'rides' column: remove commas and quotes, convert to numeric
df_ridership['rides'] = df_ridership['rides'].astype(str).str.replace(',', '')
df_ridership['rides'] = pd.to_numeric(df_ridership['rides'], errors='coerce')

# 2. Parse mixed date formats safely
df_ridership['date'] = pd.to_datetime(df_ridership['date'], format='mixed', errors='coerce')

# 3. Filter data to 2019-2021
df_ridership = df_ridership[(df_ridership['date'] >= '2019-01-01') & (df_ridership['date'] <= '2021-12-31')]

# 4. Drop any rows with nulls in key columns
df_ridership = df_ridership.dropna(subset=['route', 'date', 'rides'])

# 5. Reset index and display info
df_ridership = df_ridership.reset_index(drop=True)
print("Cleaned Ridership Shape:", df_ridership.shape)
print("Date Range:", df_ridership['date'].min(), "to", df_ridership['date'].max())
df_ridership.head()
```

```python
print("Loading weather data...")
# Skip the first 3 lines of metadata
df_weather = pd.read_csv('../data/raw/weather_chicago.csv', skiprows=3)

# 1. Simplify column names for ease of manipulation
df_weather.columns = ['date', 'temp_max', 'temp_min', 'precipitation', 'rain', 'weather_code']

# 2. Parse date column (uses DD-MM-YYYY format)
df_weather['date'] = pd.to_datetime(df_weather['date'], format='%d-%m-%Y', errors='coerce')

# 3. Filter data to 2019-2021
df_weather = df_weather[(df_weather['date'] >= '2019-01-01') & (df_weather['date'] <= '2021-12-31')]

# 4. Reset index and display info
df_weather = df_weather.reset_index(drop=True)
print("Cleaned Weather Shape:", df_weather.shape)
print("Date Range:", df_weather['date'].min(), "to", df_weather['date'].max())
df_weather.head()
```

```python
print("Loading stops data...")
df_stops = pd.read_csv('../data/raw/stops.txt')

# 1. Retain only critical spatial and identity columns
cols_to_keep = ['stop_id', 'stop_name', 'stop_lat', 'stop_lon']
df_stops = df_stops[cols_to_keep]

# 2. Remove stops with missing coordinate values
df_stops = df_stops.dropna(subset=['stop_lat', 'stop_lon'])

# 3. Reset index and display info
df_stops = df_stops.reset_index(drop=True)
print("Cleaned Stops Shape:", df_stops.shape)
df_stops.head()
```

```python
# Save clean intermediate files to processed folder
df_ridership.to_csv('../data/processed/clean_ridership.csv', index=False)
df_weather.to_csv('../data/processed/clean_weather.csv', index=False)
df_stops.to_csv('../data/processed/clean_stops.csv', index=False)

print("All cleaned datasets saved successfully in /data/processed/")
```
